# MS MARCO Data Sanity Check

This notebook performs sanity checks on the prepared MS MARCO dataset:
- Inspects corpus and query split statistics
- Loads 5 random training triplets (query, positive, BM25 hard negative) for manual review.


```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/processed")
```

```python
corpus = pd.read_parquet(DATA_DIR / "corpus.parquet")
train_q = pd.read_parquet(DATA_DIR / "train_queries.parquet")
val_q = pd.read_parquet(DATA_DIR / "val_queries.parquet")
test_q = pd.read_parquet(DATA_DIR / "test_queries.parquet")
triplets = pd.read_parquet(DATA_DIR / "train_triplets.parquet")

print(f"Corpus size: {len(corpus):,} passages")
print(f"Train queries: {len(train_q):,}")
print(f"Val queries: {len(val_q):,}")
print(f"Test queries: {len(test_q):,}")
print(f"Train triplets: {len(triplets):,}")
```

## Inspect 5 Random Training Triplets


```python
sample = triplets.sample(5, random_state=42)
for i, (_, row) in enumerate(sample.iterrows(), 1):
    print(f"=== TRIPLET {i} ===")
    print(f"Query ID: {row['qid']}")
    print(f"Query: {row['query_text']}\n")
    print(f"Positive PID: {row['pos_pid']}")
    print(f"Positive Text: {row['pos_text']}\n")
    print(f"Negative PID (BM25 Hard Negative): {row['neg_pid']}")
    print(f"Negative Text: {row['neg_text']}")
    print("-" * 70 + "\n")
```
